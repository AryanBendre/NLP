# Practical No. 1: Mini Generative AI – Next Word Prediction

## Aim

To understand the basic concept of Generative Artificial Intelligence by developing a simple next-word prediction model using Python.

## Theory

Generative Artificial Intelligence (Generative AI) is a branch of AI that creates new content such as text, images, audio, and code by learning patterns from existing data. One of the simplest text generation techniques is next-word prediction, where the system predicts the most probable next word based on previously learned word sequences. This concept is the basic building block of modern Large Language Models (LLMs) like ChatGPT.

## Algorithm

Start the program.

Create a list of sample sentences.

Split each sentence into individual words.

Store every word along with its next word in a dictionary.

Display the learned model.

Accept a starting word from the user.

Randomly select the next word from the learned dictionary.

Repeat the process until no next word is available or the desired number of words is generated.

Display the generated text.

Stop.

## Python Program

```python
import random
# Training sentences
sentences = [
```

"I love data science",

"I love machine learning",

"data science is amazing",

"machine learning is powerful",

"I enjoy learning python"

```python
]
# Dictionary to store word relationships
model = {}
# Build the model
for sentence in sentences:
words = sentence.split()
for i in range(len(words) - 1):
current_word = words[i]
next_word = words[i + 1]
if current_word not in model:
model[current_word] = []
model[current_word].append(next_word)
print("Learned Model")
print(model)
# User input
word = input("\nEnter starting word: ")
generated_text = word
# Generate text
for i in range(5):
if word in model:
word = random.choice(model[word])
generated_text = generated_text + " " + word
else:
break
print("\nGenerated Text:")
print(generated_text)
```

## Sample Output

## Conclusion

The practical demonstrated the implementation of a simple next-word prediction model using Python. It helped in understanding how Generative AI learns patterns from text and generates meaningful sentences by predicting one word at a time. This concept forms the foundation of modern text generation models.

## Python Program

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
# Load dataset
df = pd.read_csv("/content/Sample - Superstore.csv", encoding='latin1')
# Convert Order Date to datetime
df['Order Date'] = pd.to_datetime(df['Order Date'])
# Sales trend over time
df_grouped = df.groupby(df['Order Date'].dt.year)['Sales'].sum()
df_grouped.plot(kind='line', marker='o', title='Yearly Sales Trend')
plt.xlabel('Year')
plt.ylabel('Total Sales')
plt.show()
# Sales by Category
plt.figure(figsize=(6, 4))
sns.barplot(x=df['Category'], y=df['Sales'], estimator=sum, ci=None)
plt.title('Total Sales by Category')
plt.xticks(rotation=45)
plt.show()
# Profit vs. Sales scatter plot
plt.figure(figsize=(6, 4))
sns.scatterplot(x=df['Sales'], y=df['Profit'], alpha=0.5)
plt.title('Sales vs Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.show()
```

## Part A: One-Way ANOVA

## Problem Statement

Consider the exam scores of students from three different classes. Perform a One-Way ANOVA test to determine whether there is a significant difference in the mean exam scores among the classes. If significant, perform a Tukey HSD post-hoc test.

Class A: [85, 90, 88, 82, 87]
Class B: [76, 78, 80, 81, 75]
Class C: [92, 88, 94, 89, 90]

## Hypothesis

H₀ (Null Hypothesis): There is no significant difference in the mean scores of the three classes.

H₁ (Alternative Hypothesis): There is a significant difference in the mean scores of at least one class.

## Python Program

```python
import numpy as np
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd
# Data: exam scores for each class
class_A = [85, 90, 88, 82, 87]
class_B = [76, 78, 80, 81, 75]
class_C = [92, 88, 94, 89, 90]
# Perform One-Way ANOVA
```

f_statistic, p_value = stats.f_oneway(

```python
class_A, class_B, class_C
)
# Display results
print("F-Statistic:", f_statistic)
print("P-Value:", p_value)
# Draw conclusion
alpha = 0.05
if p_value < alpha:
print("Reject the Null Hypothesis: There is a significant difference in the mean exam scores.")
else:
print("Fail to Reject the Null Hypothesis: There is no significant difference in the mean exam scores.")
```

## Sample Output

## Interpretation

Since the p-value is less than 0.05, the null hypothesis is rejected. There is a significant difference between the mean scores of the classes. The Tukey test identifies which specific groups differ.

## Part C: Two-Sample T-Test

## Problem Statement

A company wants to compare the time taken to complete a task between two groups of employees.

Group I – Experience (0–1 year):

[85, 95, 100, 80, 90, 97, 104, 95, 88, 92, 94, 99]

Group II – Experience (1–2 years):

[83, 85, 96, 92, 100, 104, 94, 95, 88, 90, 93, 94]

Perform a Two-Sample T-Test.

## Hypothesis

H₀: There is no significant difference in the mean task completion times.

H₁: There is a significant difference in the mean task completion times.

## Python Program

```python
import scipy.stats as stats
import numpy as np
# Data for both groups
group1 = [85, 95, 100, 80, 90, 97, 104, 95, 88, 92, 94, 99]
group2 = [83, 85, 96, 92, 100, 104, 94, 95, 88, 90, 93, 94]
# Perform Two-Sample T-Test
```

t_stat, p_value = stats.ttest_ind(group1, group2)

```python
# Display the result
print(f"T-statistic: {t_stat}")
print(f"P-value: {p_value}")
# Hypothesis testing conclusion
alpha = 0.05
if p_value < alpha:
print("Reject the null hypothesis: There is a significant difference in the mean times.")
else:
print("Fail to reject the null hypothesis: There is no significant difference in the mean times.")
```

## Sample Output

## Interpretation

If the p-value is greater than 0.05, the null hypothesis is not rejected. Therefore, there is no statistically significant difference between the mean task completion times of the two groups.
